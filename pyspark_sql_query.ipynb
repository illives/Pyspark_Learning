```python
import os
os.environ["HADOOP_HOME"] = "C:\\hadoop"
os.environ["hadoop.home.dir"] = "C:\\hadoop"
```

```python
spark = SparkSession.builder.appName('query_sql')\
.config('spark.sql.repl.eagerEval.enabled', True).getOrCreate()
```

```python
df = spark.read.parquet('.//data/DATASETS/DATASETS/LOGINS.parquet')
df
```

```python
df2 = spark.sql('''
SELECT * FROM {tabela}
WHERE estado in {lista_estados}
''', tabela=df, lista_estados=('RR', 'MA'))

df2
```
