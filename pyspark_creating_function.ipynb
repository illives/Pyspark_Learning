```python
import pyspark.sql.functions as F
from pyspark.sql import SparkSession

from pyspark.sql.types import StringType
from pyspark.sql.functions import udf
```

```python
spark = SparkSession.builder.appName('creating_function')\
.config('spark.sql.repl.eagerEval.enabled', True).getOrCreate()

spark
```

```python
df = spark.read.parquet('./data/DATASETS/DATASETS/LOGINS.parquet')
df.summary
```

```python
@udf(returnType = StringType()) # Se usa este decorador
def nome_estado(sigla):
    if sigla == 'SP':
        return 'São Paulo'
    elif sigla == 'RJ':
        return 'Rio de Janeiro'
    else:
        return 'Outros Estados'
```

```python
df.select('estado', nome_estado('estado'))
```
